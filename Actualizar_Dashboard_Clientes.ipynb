# 🔄 Actualizar Dashboard Clientes
Actualiza las **8 secciones** del dashboard de clientes.

| Sección | Fuente |
|---|---|
| Precios RPM | `IPC TODESCA.xlsx` · `Cuadro Mensual - Capítulos Nuevo.xlsx` |
| Actividad EsAE | `Base EsAE.xlsx` |
| Reservas | `pasivos reservas.xlsx` · `Reservas brutas y depósitos.xlsx` |

> **Nota:** el dashboard clientes lee los mismos archivos `.js` que el dashboard maestro. Correr este notebook es suficiente para refrescar las secciones relevantes.


```python
import sys, os

DASHBOARD_DIR = r"C:\Users\fscalise\OneDrive - ECOGO S.A\BD\07 Tableros\EcoGo-Dashboard"
sys.path.insert(0, DASHBOARD_DIR)

import refresh as R

print(f"Dashboard: {DASHBOARD_DIR}")
print(f"Excel base: {R.BASE_EXCEL}")
print(f"Data dir:   {R.DATA_DIR}")
print()
print("Secciones a actualizar: Precios RPM · Actividad EsAE · Reservas · Monetarias · Deuda · Comercio exterior")
```

## 1 · Precios RPM

```python
status = R.Status()
d = R.extract_precios(status)
if d:
    sz = R.save_data('precios', d)
    # Mostrar solo los datos RPM
    rpm = d.get('proyeccion', {})
    chart22 = d.get('chart22', [])
    print(f"✅ precios.js — {sz:,} bytes")
    if rpm and rpm.get('filas'):
        print(f"   Proyección RPM: {len(rpm['filas'])} capítulos · {rpm.get('header_periodo','')}")
    if chart22:
        ult = [p for p in chart22 if p.get('rpm_mm') is not None]
        if ult:
            u = ult[-1]
            rpm_pct = round(u['rpm_mm']*100, 2) if u['rpm_mm'] else '—'
            print(f"   Último RPM m/m: {rpm_pct}% ({u['fecha'][:7]})")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 2 · Actividad EsAE

```python
status = R.Status()
d = R.extract_emae_series(status)
if d:
    sz = R.save_emae_series(d)
    orig = d.get('original', [])
    print(f"✅ emae_series.js — {sz:,} bytes")
    if orig:
        print(f"   Serie original: {len(orig)} observaciones · último: {orig[-1].get('date','')}")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 3 · Reservas

```python
status = R.Status()
d = R.extract_reservas(status)
if d:
    sz = R.save_data('reservas', d)
    rin = d.get('rin', {})
    g5  = d.get('g5', [])
    print(f"✅ reservas.js — {sz:,} bytes")
    if rin:
        print(f"   RIN: {len(rin.get('dates',[]))} fechas · {len(rin.get('rows',[]))} filas")
    if g5:
        print(f"   G5 (reservas diarias): {len(g5)} observaciones · último: {g5[-1].get('d','')}")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 4 · Monetarias


```python
status = R.Status()
d = R.extract_monetarias(status)
if d:
    sz = R.save_data('monetarias', d)
    c = d.get('cliente_m3_usd', {})
    print(f"✅ monetarias.js — {sz:,} bytes")
    if c.get('dates'):
        print(f"   M3 (USD CCL): {len(c['dates'])} meses · último: {c['dates'][-1]}")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 5 · Deuda


```python
status = R.Status()
d = R.extract_deuda(status)
if d:
    sz = R.save_data('deuda', d)
    lm = d.get('lopez_murphy', {})
    vm = d.get('vencimientos_mensual', {})
    rf = d.get('refinanciamiento', {})
    print(f"✅ deuda.js — {sz:,} bytes")
    if lm.get('rows'):
        print(f"   Lopez Murphy: {len(lm['rows'])} períodos")
    if vm.get('rows'):
        print(f"   Vencimientos mensual: {len(vm['rows'])} períodos")
    if rf.get('categories'):
        print(f"   Refinanciamiento: {len(rf['categories'])} categorías")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 6 · Comercio exterior
> Gráficos de `Comercio exterior.xlsx` y `Terminos del Intercambio.xlsx`, más el cuadro mensual de `Estimación Comercio Ext.xlsx` (hoja `Proy_mensual`).
> El cuadro corta solo donde arranca la proyección: las filas pintadas quedan afuera, así que cuando se despinta un mes porque ya hay dato observado, ese mes entra solo en la próxima corrida.


```python
status = R.Status()
d = R.extract_comercio(status)
if d:
    sz = R.save_data('comercio', d)
    tbl = d.get('cuadro_mensual', {})
    print(f"✅ comercio.js — {sz:,} bytes")
    for k, lab in [('impo_desest', 'Impo desestacionalizadas'),
                   ('saldos', 'Saldos comerciales 12M'),
                   ('terminos', 'Términos de intercambio')]:
        b = d.get(k, {})
        if b.get('dates'):
            print(f"   {lab}: {len(b['dates'])} meses · último: {b['dates'][-1][:7]}")
    if tbl.get('rows'):
        print(f"   Cuadro mensual: {len(tbl['rows'])} meses · hasta {tbl['rows'][-1][0]} (sin proyección)")
for r in status.results:
    print(f"  [{r[0]}] {r[1]}: {r[2]}")
```

## 8 · Internacional · Consensus (LatinFocus)
> Requiere tener el PDF LatinFocus del mes correspondiente. Correr manualmente desde terminal.


```python
import subprocess, os

# Ajustar al nombre del PDF del mes actual
PDF_NAME = 'LatinFocus Consensus Forecast - May 2026.pdf'
PDF_PATH = os.path.join(DASHBOARD_DIR, PDF_NAME)

if not os.path.exists(PDF_PATH):
    print(f'⚠️  PDF no encontrado: {PDF_PATH}')
    print('   Copiá el PDF a la carpeta del dashboard y actualizá PDF_NAME.')
else:
    result = subprocess.run(
        ['python', os.path.join(DASHBOARD_DIR, 'parse_latinfocus.py'), PDF_PATH],
        capture_output=True, text=True, cwd=DASHBOARD_DIR
    )
    print(result.stdout)
    if result.returncode != 0:
        print('❌ Error:', result.stderr)
```
